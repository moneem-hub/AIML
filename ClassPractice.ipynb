```python
import pandas as pd

students = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
        "department": ["CS", "IT", "CS", "IT", "CS"],
        "age": [20, 22, 21, 23, 25]
    }
)

scores = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 6],
        "marks": [85, 90, 78, 88]
    }
)

print("Students:")
print(students)
print("\nScores:")
print(scores)
```

```python
pd.merge(
    students,
    scores,
    on= "student_id",
    how = "inner"
)
```

```python
import pandas as pd

students = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
        "department": ["CS", "IT", "CS", "IT", "CS"],
        "age": [20, 22, 21, 23, 25]
    }
)

scores = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 6],
        "marks": [85, 90, 78, 88]
    }
)

# Corrected merge
right_joined_df = pd.merge(students, scores, on="student_id", how="right")
print(right_joined_df)
```

```python
pd.merge(
    students,
    scores,
    on= "student_id",
    how = "left"
)
```

```python
students_indeexed = students.set_index("student_id")
scores_indexed = scores.set_index("student_id")

result = students_indeexed.join(scores_indexed, how="left")
print(result)
```

```python
df1 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "marks": [85, 90]
    }
)

df2 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "age": [20, 22]
    }
)

result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
left = pd.DataFrame({
    "name": ["Ali", "Sara"]
})
right = pd.DataFrame({
    "marks": [85, 90]
})

result = pd.concat([left, right], axis=1)
print(result)
```

```python
df1 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "marks": [85, 90]
    }
)

df2 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "age": [20, 22]
    }
)

result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)
```

```python
import pandas as pd

df = pd.DataFrame(
    {
        "month": ["January", "February", "March", "April", "May"],
        "sales": [100, 150, 130, 200, 250],
        "profit": [20, 30, 25, 40, 50]
    }
)

print(df)
```

```python
import matplotlib.pyplot as plt
df.plot(
    x="month",
    y="sales",
    kind="area"
)

plt.show()
```

```python
df.plot(x="month", y="sales", kind="bar", title="Monthly Sales")
plt.show()
```

```python
df.plot(x="month", y="sales", kind="line", marker='o', title="Monthly Sales Trend")
plt.show()
```

```python
df["sales"].plot(kind="hist", bins=5, title="Sales Distribution")
plt.show()
```

```python
df["sales"].plot(kind = "density")
plt.show()
```

```python
df["sales"].plot(kind = "kde")
plt.show()
```

```python
df[["sales", "profit"]].plot(kind="box")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style="whitegrid")
```

```python
df=pd.DataFrame({
    "department":["CS","IT","CS","IT","CS","IT","CS","IT"],
    "gender":["M","F","M","F","F","M","M","F"],
    "marks":[85,90,78,95,88,70,92,82],
    "hours":[4,6,3,7,5,2,6,4],
    "age":[20,22,21,23,25,24,22,21]
})
sns.scatterplot(data=df, x="hours", y="marks", hue="department", s=100)
plt.show()
```

```python
sns.displot(df["marks"])
plt.show()
```

```python
sns.jointplot(
    data=df,
    x = "hours",
    y = "marks",
    kind = "scatter"
)

plt.show()
```

```python
sns.pairplot(
    df,
    vars=["hours", "marks", "age"],
    hue="department"
)

plt.show()
```

```python
sns.rugplot(
    data= df,
    x = "marks"
)

plt.show()
```

```python
sns.boxplot(
    data=df,
    x="department",
    y="marks"
)

plt.show()
```

```python
#ViolintPlot
sns.violinplot(
    data=df,
    x="department",
    y="marks"
)

plt.show()
```

```python
#Stripplot
sns.stripplot(
    data=df,
    x="department",
    y="marks",
    jitter=True
)
plt.show()
```

```python
#Swarm plot
sns.swarmplot(
    data=df,
    x="department",
    y="marks"
)

plt.show()
```

```python
# Countplot
sns.countplot(
    data=df,
    x="department"
)

plt.show()
```
