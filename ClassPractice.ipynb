```python
import pandas as pd

students = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
        "department": ["CS", "IT", "CS", "IT", "CS"],
        "age": [20, 22, 21, 23, 25]
    }
)

scores = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 6],
        "marks": [85, 90, 78, 88]
    }
)

print("Students:")
print(students)
print("\nScores:")
print(scores)
```

```python
pd.merge(
    students,
    scores,
    on= "student_id",
    how = "inner"
)
```

```python
import pandas as pd

students = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
        "department": ["CS", "IT", "CS", "IT", "CS"],
        "age": [20, 22, 21, 23, 25]
    }
)

scores = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 6],
        "marks": [85, 90, 78, 88]
    }
)

# Corrected merge
right_joined_df = pd.merge(students, scores, on="student_id", how="right")
print(right_joined_df)
```

```python
pd.merge(
    students,
    scores,
    on= "student_id",
    how = "left"
)
```

```python
students_indeexed = students.set_index("student_id")
scores_indexed = scores.set_index("student_id")

result = students_indeexed.join(scores_indexed, how="left")
print(result)
```

```python
df1 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "marks": [85, 90]
    }
)

df2 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "age": [20, 22]
    }
)

result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
left = pd.DataFrame({
    "name": ["Ali", "Sara"]
})
right = pd.DataFrame({
    "marks": [85, 90]
})

result = pd.concat([left, right], axis=1)
print(result)
```

```python
df1 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "marks": [85, 90]
    }
)

df2 = pd.DataFrame(
    {
        "name": ["Ali", "Sara"],
        "age": [20, 22]
    }
)

result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)
```

```python
df = pd.DataFrame(
    {
        "department": ["CS", "IT", "CS", "IT", "CS"],
        "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
        "marks": [85, 90, 78, 88, 92],
        "ages": [20, 22, 21, 23, 25]
    }
)

df.groupby("department")["marks"].mean()
```

```python
df.groupby("department")["marks"].agg(
    ["mean", "max", "min", "count"]
)
```

```python
df = pd.DataFrame(
    {
        "name": ["Ali", "Sara", "Sara", "Ahmed"],
        "age": [18, 25, 42, 65]
    }
)

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 20, 40, 60, 100],
    labels=["Young", "Adult","Middle-aged", "Senior"]
)
```
