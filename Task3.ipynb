# BOW (Bag of words)

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample Data
corpus = [
    "The cat sat on the mat",
    "The dog sat on the log",
    "The cat and the dog played"
]

# Initialize BoW
vectorizer = CountVectorizer()

# Transform text to matrix
bow_matrix = vectorizer.fit_transform(corpus)

print("Feature Names (Vocabulary):", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow_matrix.toarray())
```

# TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF
tfidf_vectorizer = TfidfVectorizer()

# Transform text to matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

print("TF-IDF Matrix:\n", tfidf_matrix.toarray())
```

# N-Gram

```python
from sklearn.feature_extraction.text import CountVectorizer

# Unigram Model(Deafult)
bow_unigram = CountVectorizer(ngram_range=(1, 1))
print("Unigram Features:", bow_unigram.fit(corpus).get_feature_names_out())

# Bigram Model
bow_bigram = CountVectorizer(ngram_range=(2, 2))
print("Bigram Features:", bow_bigram.fit(corpus).get_feature_names_out())
```

# Word Embedings

```python
import gensim.downloader as api
import numpy as np

# Load pre-trained Word2Vec model
model_w2v = api.load("word2vec-google-news-300")

# Get vector for a word
word_vector = model_w2v['King']
print("Shape of 'King' vector:", word_vector.shape)

# Find similar words
similar_words = model_w2v.most_similar('King', topn=5)
print("Words similar to 'King':", similar_words)

# Average Word Vectors for a sentence
def get_sentence_vector(sentence, model):
    words = sentence.lower().split()
    vectors = [model[word] for word in words if word in model]
    if vectors:
        return np.mean(vectors, axis=0)
    return np.zeros(model.vector_size)

# Example sentence
sent_vec = get_sentence_vector("King is strong", model_w2v)
print("Sentence vector shape:", sent_vec.shape)
```
