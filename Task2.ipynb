# Withou Vectorization

```python
# Training data
X = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Hyperparameters
w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000

m = len(X)

for epoch in range(epochs):
    # Initialize gradients
    dw = 0.0
    db = 0.0

    # Calculate gradients using a loop (SLOW)
    for i in range(m):
        y_pred = w * X[i] + b
        error = y_pred - y[i]
        
        dw += error * X[i]
        db += error

    # Average gradients
    dw = dw / m
    db = db / m

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight (Loop):", w)
print("Bias (Loop):", b)

# Prediction
new_x = 6
prediction = w * new_x + b
print("Predicted value (Loop):", prediction)
```

# With Vectorization

```python
import numpy as np

# Training data (Convert to NumPy arrays)
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Hyperparameters
w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000

m = len(X)

for epoch in range(epochs):
    # Vectorized Prediction: y_pred = w * X + b (Calculates for all X at once)
    y_pred = w * X + b
    
    # Vectorized Error: error = y_pred - y (Calculates for all elements at once)
    error = y_pred - y
    
    # Vectorized Gradients: 
    # dw = sum(error * X) / m
    # db = sum(error) / m
    dw = np.sum(error * X) / m
    db = np.sum(error) / m
    
    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight (Vectorized):", w)
print("Bias (Vectorized):", b)

# Prediction
new_x = 6
prediction = w * new_x + b
print("Predicted value (Vectorized):", prediction)
```
