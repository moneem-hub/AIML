```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

current = node1  # Start at the head of the list
while current is not None:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```

```python
## Single Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Build: 10 -> 20 -> 30 -> None
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

# Traverse it
current = head
while current is not None:
    print(current.data)
    current = current.next
```

```python
# Doubly Linked List
class DoublyNode:
    def __init__(self, data):
        self.data = data
        self.next = None

node1 = DoublyNode(10)
node2 = DoublyNode(20)
node3 = DoublyNode(30)

node1.next = node2
node2.prev = node1
node2.next = node3
node3.prev = node2

# Traverse forward
current = node1
while current is not None:
    print(current.data)
    current = current.next

# Traverse Backward from end
current = node2
while current is not None:
    print(current.data)
    current = current.prev
```

```python
# Circular Linked List
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3
node3.next = node1

current = node1
count = 0
while count < 6:
    print(current.data)
    current = current.next
    count += 1
```
