```python
numbers = [5, 2, 8, 1, 9]
len(numbers)          # 5
numbers.sort()         # [1, 2, 5, 8, 9] — modifies in place
numbers.reverse()      # reverses in place
5 in numbers            # True — membership check
```

```python
#List Comprehension
# The long way:
squares = []
for x in range(1, 6):
    squares.append(x ** 2)

print(squares)
```

```python
squares = []
# The comprehension way — same result, one line:
squares = [x ** 2 for x in range(1, 6)]
print(squares)
```

```python
even = [x for x in range(10) if x % 2 == 0]
print(even)
```

```python
cube = [x ** 3 for x in range(1,6) if x % 2 != 0] #Print Cube for 1 to 6 odd numbers
print(cube)
```
