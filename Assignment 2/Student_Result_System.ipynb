## Setup: constants and helper functions
These helpers are used by all the sections. They check the input, so wrong values never crash the program.

````python
import keyword
import math

# Constants used by the whole notebook
SUBJECTS = ["Mathematics", "Physics", "Chemistry", "English", "Computer Science"]
MAX_MARKS = 100               # maximum marks in one subject
PASS_PERCENTAGE = 50          # below this percentage the student fails
DISTINCTION_PERCENTAGE = 75   # distinction rule chosen for this notebook (75% or more)


def parse_mark(text):
    """Convert the text typed by the user into a valid mark (float).
    Raises ValueError with a clear message if the text is not a valid mark."""
    text = text.strip()                      # remove spaces around the input
    if text == "":
        raise ValueError("The marks cannot be empty.")
    mark = float(text)                       # type casting: str -> float (ValueError if not a number)
    if not math.isfinite(mark):              # rejects 'nan' and 'inf', which float() accepts
        raise ValueError("The marks must be a real number.")
    if mark < 0 or mark > MAX_MARKS:
        raise ValueError(f"The marks must be between 0 and {MAX_MARKS}.")
    return mark


def read_mark(subject, source=input):
    """Keep asking for the marks of one subject until the user enters a valid value
    or chooses to stop by typing 'q'. Returns (raw_text, mark), or None if stopped."""
    while True:                                           # while loop: repeat until valid or quit
        raw = source(f"Enter marks for {subject} (0-{MAX_MARKS}, or 'q' to stop): ")
        if raw.strip().lower() == "q":
            return None                                   # user chose to stop
        try:
            return raw.strip(), parse_mark(raw)           # valid value: leave the loop
        except ValueError as error:
            print(f"  Invalid input ({error}) Please try again.")


def get_student_name(source=input):
    """Ask for the student's name until a sensible name is entered."""
    while True:
        raw = source("Enter the student's name: ")
        name = raw.strip()
        if name == "":
            print("  Invalid input (the name cannot be empty). Please try again.")
        elif any(character.isdigit() for character in name):
            print("  Invalid input (the name cannot contain digits). Please try again.")
        else:
            return raw                                    # returned exactly as typed


def collect_marks(subjects, source=input):
    """Collect one valid mark for every subject. Returns (raw_texts, marks) or None."""
    raw_texts, marks = [], []
    for subject in subjects:
        result = read_mark(subject, source)
        if result is None:
            return None
        raw_texts.append(result[0])
        marks.append(result[1])
    return raw_texts, marks


def make_fake_input(values):
    """Behaves like input() but reads from a list (used to demonstrate and test the loops)."""
    iterator = iter(values)
    def fake_input(prompt=""):
        value = next(iterator)
        print(f"{prompt}{value}")
        return value
    return fake_input


def calculate_results(marks):
    """Return (total, average, percentage) for a list of marks."""
    total = sum(marks)
    average = total / len(marks)
    percentage = total / (len(marks) * MAX_MARKS) * 100
    return total, average, percentage


def get_grade(percentage):
    """Grade rules from the question (see section d)."""
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "Fail"


print("Setup complete.")
````

### d. 1 Grade with `if` / `elif` / `else`

````python
total_marks, average_marks, percentage = calculate_results(marks)

if percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "Fail"

print(f"Percentage {percentage:.2f}%  ->  {'Grade ' + grade if grade != 'Fail' else 'Fail'}")

print("\nChecking the boundaries:")
for value in [95, 80, 79.9, 70, 69.9, 60, 59.9, 50, 49.9]:
    result = get_grade(value)
    print(f"  {value:>5}%  ->  {'Grade ' + result if result != 'Fail' else 'Fail'}")
````

### d. 2 Nested conditional: passed? and distinction?

````python
if percentage >= PASS_PERCENTAGE:                  # outer check: has the student passed?
    if percentage >= DISTINCTION_PERCENTAGE:       # inner check: does the student qualify for distinction?
        result = "Passed with Distinction"
    else:
        result = "Passed (no distinction)"
else:
    result = "Failed"

print("Result:", result)
````

````python
eligibility = "Eligible to pass" if percentage >= PASS_PERCENTAGE else "Not eligible to pass"
print("Ternary expression result:", eligibility)
````

### d. 4 Why the order of conditions matters (demonstration)

````python
def grade_wrong_order(value):
    if value >= 50:            # this catches everything from 50 upward
        return "D"
    elif value >= 60:          # never reached for 60 or more
        return "C"
    elif value >= 70:
        return "B"
    elif value >= 80:
        return "A"
    else:
        return "Fail"

print("85% with the correct order       -> Grade", get_grade(85))
print("85% with the conditions reversed -> Grade", grade_wrong_order(85), " (wrong!)")
````

### Explanation: `if`, `elif` and `else`
* **`if`** starts the decision. Its block runs when its condition is `True`.
* **`elif`** means "else if". It is checked only when all earlier conditions were `False`. There can be many `elif` blocks.
* **`else`** is the last block. It runs when none of the conditions above was `True`. It has no condition and is optional.

Only the **first** block whose condition is `True` runs. The rest are skipped.

### Explanation: why the order of conditions matters in a grading system
Python checks the conditions from top to bottom and stops at the first `True` one. A mark of 85 is also `>= 70`, `>= 60` and `>= 50`, so the **highest grade must be tested first** (`>= 80`, then `>= 70`, and so on). If `>= 50` were tested first, every passing student would receive Grade D, as the demonstration above shows.

# g. Code Analysis and Debugging
Code given in the question:
```python
marks = [45, 80, 32, 90, 65]

passed = []

for mark in marks:
    if mark < 50:
        continue
    passed.append(mark)

print(passed)
print(len(passed))
print(sum(passed) / len(passed))
```

### g. 1 What is the output of the program?

````python
marks = [45, 80, 32, 90, 65]

passed = []

for mark in marks:
    if mark < 50:
        continue
    passed.append(mark)

print(passed)
print(len(passed))
print(sum(passed) / len(passed))
````

**Output:** `[80, 90, 65]`, then `3`, then `78.33333333333333` (80 + 90 + 65 = 235, and 235 / 3).

### g. 2 Why is `continue` used?
When a mark is below 50 (45 and 32), `continue` skips the rest of that round, so `append()` is not executed and the failing mark is **not added** to the list. The loop then carries on with the next mark.

### g. 3 What would change if `continue` were replaced with `break`?

````python
passed_break = []
for mark in marks:
    if mark < 50:
        break                  # stops the whole loop at the first mark below 50
    passed_break.append(mark)

print(passed_break)
print(len(passed_break))
try:
    print(sum(passed_break) / len(passed_break))
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
````

The loop would end completely at the **first** mark below 50. The first mark is 45, so the loop stops immediately and `passed` stays empty (the marks 80, 90 and 65 are never checked). The program prints `[]` and `0`, and then crashes with `ZeroDivisionError` because `len(passed)` is 0.

### g. 4 What if all marks were below 50? Problem and solution

````python
low_marks = [45, 32, 40]
passed_low = []
for mark in low_marks:
    if mark < 50:
        continue
    passed_low.append(mark)

print("passed =", passed_low, "| len(passed) =", len(passed_low))

# The problem: dividing by len(passed), which is 0
try:
    print(sum(passed_low) / len(passed_low))
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error, "  <- the problem")

# The solution: check that the list is not empty before dividing
if passed_low:
    print(sum(passed_low) / len(passed_low))
else:
    print("Fixed: no marks of 50 or above, so the average is not calculated.")
````

**Problem:** `passed` stays empty, so `len(passed)` is 0 and `sum(passed) / len(passed)` raises `ZeroDivisionError` (division by zero).

**Solution:** check the list before dividing:
```python
if passed:
    print(sum(passed) / len(passed))
else:
    print("No passing marks")
```

### g. 5 Rewrite the loop using list comprehension

````python
passed_lc = [mark for mark in marks if mark >= 50]

print(passed_lc)
print(len(passed_lc))
print(sum(passed_lc) / len(passed_lc) if passed_lc else "No passing marks")
print("Same result as the loop version:", passed_lc == passed)
````

# Testing: four different sets of marks and invalid input
The program is tested with **four** sets of marks (more than the three required), including a boundary case. Each result is compared with the expected grade.

````python
def generate_report(name, marks):
    """Print a result report for one student and return the key results."""
    total, average, percentage = calculate_results(marks)
    grade = get_grade(percentage)
    if percentage >= PASS_PERCENTAGE:
        status = "Passed with Distinction" if percentage >= DISTINCTION_PERCENTAGE else "Passed"
    else:
        status = "Failed"
    best_index = marks.index(max(marks))
    print(f"  Student     : {name}")
    for number, (subject, mark) in enumerate(zip(SUBJECTS, marks), start=1):
        print(f"    {number}. {subject:<18}{mark:>6.1f}")
    print(f"  Total       : {total:.1f} / {len(marks) * MAX_MARKS}")
    print(f"  Average     : {average:.2f}")
    print(f"  Percentage  : {percentage:.2f}%")
    print(f"  Grade       : {'Grade ' + grade if grade != 'Fail' else 'Fail'}")
    print(f"  Status      : {status}")
    print(f"  Best subject: {SUBJECTS[best_index]} ({marks[best_index]})")
    return {"name": name, "percentage": percentage, "grade": grade, "status": status}


test_cases = [
    ("Test 1 - high scorer",                 "Ayesha Khan", [92, 88, 79, 85, 90], "A"),
    ("Test 2 - average student",             "Bilal Ahmed", [65, 58, 72, 61, 55], "C"),
    ("Test 3 - failing student",             "Usman Ali",   [35, 42, 28, 45, 39], "Fail"),
    ("Test 4 - boundary case (exactly 50%)", "Hina Raza",   [50, 50, 50, 50, 50], "D"),
]

records = []                                       # all test records, stored in a list of dictionaries
all_ok = True
for title, student_name, student_marks, expected_grade in test_cases:
    print(f"\n{title}")
    record = generate_report(student_name, student_marks)
    records.append(record)
    ok = record["grade"] == expected_grade
    all_ok = all_ok and ok
    print(f"  Expected grade: {expected_grade} | Actual grade: {record['grade']} | {'OK' if ok else 'MISMATCH'}")

print("\nSummary of all test records")
print(f"  {'Name':<14}{'Percentage':>11}{'Grade':>7}  Status")
for record in records:
    print(f"  {record['name']:<14}{record['percentage']:>10.2f}%{record['grade']:>7}  {record['status']}")
print("\nAll grade checks passed:", all_ok)
````
